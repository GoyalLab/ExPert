# Figure S5 — single CRISPRa and extended double-perturbation recovery (related to Figure 3)

Renders panels A–F from `data/` (written by `figS5_data.py`, `figS5.sbatch` STEP=data); drawn as Fig. 3 H/I (`fig3_doubles_plot.panel`) with `fig4_pb_plot`'s style. Panels go to `render/`, source tables to `render/src/`.

```python
import sys
from pathlib import Path
_paper = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'scripts' / 'common').is_dir())
sys.path.insert(0, str(_paper / 'scripts' / 'common'))
import nbtools as nb                      # puts paper/scripts/*/ on sys.path

vs = nb.vizstyle()
DATA, INPUTS, OUT, SRC = nb.folders()     # data/ (cache), inputs/, render/, render/src/
```

```python
import json
import pandas as pd
import figS5_plot as P
import fig3_doubles_plot as P3
import fig4_pb_plot as fp

fp.setup(str(nb.STYLE))
P3.plt, P3.matplotlib = fp.plt, fp.matplotlib
fp.SRC_DIR = str(SRC)
C = pd.read_csv(DATA / 'figS5_doubles_curves.csv')
meta = json.load(open(DATA / 'figS5_meta.json'))
Cs = pd.read_csv(DATA / 'figS5_singles_curves.csv')
smeta = json.load(open(DATA / 'figS5_singles_meta.json'))
```

## A — single CRISPRa perturbations: target in the top k of 8,347 candidates

```python
P3.panel(Cs, smeta, 'gain', 'scorable', 'single', OUT, '', name='figS5a_singles_gain')
nb.show(OUT, 'figS5a_singles_gain')
```

## B — (A) per screen

```python
P.by_screen(Cs, smeta, 'gain', 'scorable', 'single', OUT, 'figS5b_singles_gain_by_screen')
nb.show(OUT, 'figS5b_singles_gain_by_screen')
```

## C — loss-of-function doubles, both genes in the top k of each screen's own double genes

```python
P3.panel(C, meta, 'loss', 'in_panel', 'both', OUT, '', name='figS5c_doubles_loss')
nb.show(OUT, 'figS5c_doubles_loss')
```

## D — (C) per screen

```python
P.by_screen(C, meta, 'loss', 'in_panel', 'both', OUT, 'figS5d_doubles_loss_by_screen')
nb.show(OUT, 'figS5d_doubles_loss_by_screen')
```

## E — gain-of-function doubles, as in (C)

```python
P3.panel(C, meta, 'gain', 'in_panel', 'both', OUT, '', name='figS5e_doubles_gain')
nb.show(OUT, 'figS5e_doubles_gain')
```

## F — (E) per screen

```python
P.by_screen(C, meta, 'gain', 'in_panel', 'both', OUT, 'figS5f_doubles_gain_by_screen')
nb.show(OUT, 'figS5f_doubles_gain_by_screen')
```
