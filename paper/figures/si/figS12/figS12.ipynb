# Figure S12 — ExPert scores of the bulk RNA-seq livers, and Figure 5 split by condition (related to Figure 5)

A–B render from the cached tables in `data/` (written by `figS12_data.py`). C–H are Figure 5's own plot functions on
Figure 5's data (`inputs/fig5_final` = Figure 5's `data/final`, plus the snRNA-seq pseudobulk DE table, gene-set files,
FISH frames and nucleus measurements linked in `inputs/`), with AC and AH shown separately. Panels go to `render/`,
source tables to `render/src/`.

```python
import sys
from pathlib import Path
_paper = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'scripts' / 'common').is_dir())
sys.path.insert(0, str(_paper / 'scripts' / 'common'))
import nbtools as nb                      # puts paper/scripts/*/ on sys.path

vs = nb.vizstyle()
DATA, INPUTS, OUT, SRC = nb.folders()     # data/ (cache), inputs/, render/, render/src/
```

```python
import json
import figS12_plot as P
import figS12c_split_plot as S

P.style(vs)
meta = json.loads((DATA / 'figS12_meta.json').read_text())
a = S.parser().parse_args(['--fig5-data', str(INPUTS / 'fig5_final'), '--deg', str(INPUTS / 'ald_snrna_pseudobulk_deg.csv.gz'),
                           '--modules', str(INPUTS / 'readout_panels.tsv'),
                           '--reactome-gmt', str(INPUTS / 'Enrichr.Reactome_Pathways_2024.gmt'),
                           '--fish-frames', str(INPUTS / 'fish_frames.csv'), '--nuclei', str(INPUTS / 'nuclei_measurements.csv'),
                           '--out-dir', str(OUT), '--src-dir', str(SRC), '--vizstyle-root', str(nb.STYLE)])
c = S.setup(a, vs)                        # Fig. 5 tables + gene sets; sets fig5_final.SRC = render/src
```

## A — mean predicted-loss z of the mRNA export families, snRNA-seq vs bulk livers

```python
P.panel_a(DATA, OUT, vs, meta)
nb.show(OUT, 'figS12a_families')
```
