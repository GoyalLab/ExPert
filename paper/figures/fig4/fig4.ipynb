# Figure 4 — drug and protein-depletion responses read through CRISPR-trained prototypes

Renders panels A–H and J from the cached tables in `data/` with `fig4_plot.py`. Panel I is a static image (see below). Panels go to `render/`, their source tables to `render/src/`.

```python
import sys
from pathlib import Path
_paper = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'scripts' / 'common').is_dir())
sys.path.insert(0, str(_paper / 'scripts' / 'common'))
import nbtools as nb                      # puts paper/scripts/*/ on sys.path

vs = nb.vizstyle()
DATA, INPUTS, OUT, SRC = nb.folders()     # data/ (cache), inputs/, render/, render/src/
```

```python
import fig4_plot as P

P.style(vs)
P.SRC_DIR = SRC          # source tables -> render/src, never into data/
```
